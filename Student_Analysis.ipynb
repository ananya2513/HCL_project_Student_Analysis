```python
# PROJECT: STUDENT CAREER SUCCESS ANALYSIS SYSTEM

# Importing Libraries
import pandas as pd
import numpy as np

# Loading Dataset
df = pd.read_csv("student_career_success_dataset.csv")

# Main Program
while True:

    print("\n==========================================")
    print("     STUDENT CAREER SUCCESS ANALYSIS")
    print("==========================================")

    print("1. Overall Placement Statistics")
    print("2. Academic Performance Analysis")
    print("3. Placement Analysis by Major")
    print("4. Internship vs Placement Analysis")
    print("5. Technical and Soft Skills Analysis")
    print("6. Salary Analysis")
    print("7. Employability Score Analysis")
    print("8. Search Student Record")
    print("9. Overall Dataset Summary")
    print("10. Exit")

    choice = input("\nEnter your choice (1-10): ")

    # OPTION 1: PLACEMENT STATISTICS
    if choice == '1':

        total = len(df)

        placed = (df["Placement_Status"] == "Placed").sum()

        not_placed = total - placed

        percentage = np.mean(
            df["Placement_Status"] == "Placed"
        ) * 100

        print("\n----- PLACEMENT STATISTICS -----")
        print("Total Students:", total)
        print("Placed Students:", placed)
        print("Not Placed Students:", not_placed)
        print("Placement Percentage:", round(percentage, 2), "%")

    # OPTION 2: ACADEMIC ANALYSIS
    elif choice == '2':

        print("\n----- ACADEMIC PERFORMANCE -----")

        print("Average CGPA:",
              round(df["CGPA"].mean(), 2))

        print("Average Attendance:",
              round(df["Attendance_Percentage"].mean(), 2), "%")

        print("Average Study Hours:",
              round(df["Study_Hours_Per_Week"].mean(), 2))

        print("\nAcademic Performance Categories:")
        print(df["Academic_Performance"].value_counts())

    # OPTION 3: PLACEMENT BY MAJOR
    elif choice == '3':

        print("\n----- PLACEMENT BY MAJOR -----")

        result = df.groupby("Major").agg(
            Total_Students=("Student_ID", "count"),
            Placed_Students=(
                "Placement_Status",
                lambda x: (x == "Placed").sum()
            )
        )

        result["Placement_Percentage"] = (
            result["Placed_Students"] /
            result["Total_Students"] * 100
        )

        print(result.round(2))

    # OPTION 4: INTERNSHIP ANALYSIS
    elif choice == '4':

        print("\n----- INTERNSHIP ANALYSIS -----")

        result = df.groupby("Internships").agg(
            Total_Students=("Student_ID", "count"),
            Placed_Students=(
                "Placement_Status",
                lambda x: (x == "Placed").sum()
            )
        )

        result["Placement_Percentage"] = (
            result["Placed_Students"] /
            result["Total_Students"] * 100
        )

        print(result.round(2))

    # OPTION 5: SKILLS ANALYSIS
    elif choice == '5':

        print("\n----- SKILLS ANALYSIS -----")

        print("Average Programming Skill:",
              round(df["Programming_Skill"].mean(), 2))

        print("Average Communication Skills:",
              round(df["Communication_Skills"].mean(), 2))

        print("Average Teamwork:",
              round(df["Teamwork"].mean(), 2))

        print("Average Problem Solving:",
              round(df["Problem_Solving"].mean(), 2))

        print("Average Projects Completed:",
              round(df["Projects_Completed"].mean(), 2))

        print("Average Certifications:",
              round(df["Certifications"].mean(), 2))

    # OPTION 6: SALARY ANALYSIS
    elif choice == '6':

        print("\n----- SALARY ANALYSIS -----")

        placed_data = df[
            df["Placement_Status"] == "Placed"
        ]

        salary = placed_data["Starting_Salary_USD"]

        print("Average Starting Salary:",
              round(salary.mean(), 2), "USD")

        print("Highest Starting Salary:",
              salary.max(), "USD")

        print("Lowest Starting Salary:",
              salary.min(), "USD")

        print("\nAverage Salary by Company Tier:")

        print(
            placed_data.groupby("Company_Tier")[
                "Starting_Salary_USD"
            ].mean().round(2)
        )

    # OPTION 7: EMPLOYABILITY ANALYSIS
    elif choice == '7':

        print("\n----- EMPLOYABILITY ANALYSIS -----")

        score = df["Employability_Score"]

        print("Average Score:", round(score.mean(), 2))

        print("Highest Score:", round(score.max(), 2))

        print("Lowest Score:", round(score.min(), 2))

        print("Median Score:", round(np.median(score), 2))

        print("90th Percentile:", round(np.percentile(score, 90), 2))

    # OPTION 8: SEARCH STUDENT RECORD
    elif choice == '8':

        print("\n----- STUDENT RECORD SEARCH -----")

        student_id = input("Enter Student ID (e.g. ST000001): ")

        record = df[
            df["Student_ID"].str.upper() == student_id.upper()
        ]

        if not record.empty:

            print("\nStudent Details:")
            print(record.T)

        else:

            print("Student ID not found!")

    # OPTION 9: DATASET SUMMARY
    elif choice == '9':

        print("\n----- DATASET SUMMARY -----")

        print("Total Records:", df.shape[0])

        print("Total Columns:", df.shape[1])

        print("\nAvailable Majors:")
        print(df["Major"].unique())

        print("\nAvailable Career Fields:")
        print(df["Career_Field"].unique())

        print("\nDataset Information:")
        df.info()

    # OPTION 10: EXIT
    elif choice == '10':

        print("\nThank you for using the Student Career")
        print("Success Analysis System!")

        break

    else:

        print("\nInvalid Choice! Please enter a number from 1 to 10.")
```
